```python
def total_calc(bill_amount, tip_perc=15):
    """
    Calculates the tip and total amount to be paid at a restaurant,
    prints a detailed breakdown, and returns the total amount.

    Parameters:
    bill_amount (float): The base cost of the meal before tip.
    tip_perc (float): The tip percentage to apply (default is 15%).

    Returns:
    float: The final total amount including the tip.
    """
    # Calculate the tip amount and round to 2 decimal places
    tip_amount = bill_amount * (tip_perc / 100)
    total_amount = bill_amount + tip_amount

    # Print out the formatted payment breakdown
    print("\n========================================")
    print("           RESTAURANT BILL              ")
    print("========================================")
    print(f"Base Bill Amount : ₹{bill_amount:>10.2f}")
    print(f"Tip ({tip_perc}%)       : ₹{tip_amount:>10.2f}")
    print("----------------------------------------")
    print(f"Total Amount Due : ₹{total_amount:>10.2f}")
    print("========================================\n")

    return round(total_amount, 2)


def main():
    print("Welcome to the Restaurant Tip Calculator!")

    # Get valid bill amount from the user
    while True:
        try:
            bill = float(input("Enter the bill amount: ₹"))
            if bill > 0:
                break
            print("Bill amount must be greater than zero.")
        except ValueError:
            print("Invalid input. Please enter a numeric value (e.g., 850.50).")

    # Get valid tip percentage from the user (or use default)
    while True:
        tip_input = input("Enter tip percentage (press Enter for default 15%): ").strip()
        if not tip_input:
            tip_percentage = 15.0
            break
        try:
            tip_percentage = float(tip_input)
            if tip_percentage >= 0:
                break
            print("Tip percentage cannot be negative.")
        except ValueError:
            print("Invalid input. Please enter a valid number for the percentage.")

    # Call total_calc() with the user's arguments
    final_bill = total_calc(bill, tip_percentage)
    print(f"Please pay ₹{final_bill:.2f}. Thank you for dining with us!")


# Execute the program
if __name__ == "__main__":
    main()
```
