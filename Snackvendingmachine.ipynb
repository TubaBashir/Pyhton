```python
def display_snacks():
    """Displays available snacks and returns the menu dictionary."""
    snacks = {
        "1": ("Masala Chips", 20),
        "2": ("Chocolate Bar", 35),
        "3": ("Salted Peanuts", 15),
        "4": ("Fruit Juice", 30),
        "5": ("Oatmeal Cookies", 25)
    }

    print("========================================")
    print("         SMART SNACK VENDING            ")
    print("========================================")
    print("Code | Snack                  | Price")
    print("-----+------------------------+---------")
    for code, (name, price) in snacks.items():
        print(f" {code:<3} | {name:<22} | ₹{price}")
    print("========================================")
    print("Accepted Coins: ₹1, ₹2, ₹5, ₹10, ₹20\n")
    return snacks


def calculate_change(total_inserted, item_price):
    """Calculates and returns the change owed to the user."""
    return total_inserted - item_price


def dispense_snack(snack_name, item_price, total_inserted, change_owed):
    """Prints the final dispensing receipt and change breakdown."""
    print("\n========================================")
    print("           DISPENSING SNACK             ")
    print("========================================")
    print(f"Snack Selected  : {snack_name}")
    print(f"Item Price      : ₹{item_price}")
    print(f"Total Inserted  : ₹{total_inserted}")
    print(f"Change Returned : ₹{change_owed}")
    print("----------------------------------------")
    print(f"Enjoy your {snack_name}!")
    print("========================================\n")


def main():
    # Valid Indian coin denominations in Rupees
    VALID_COINS = [1, 2, 5, 10, 20]
    snacks = display_snacks()

    # 1. Select a valid snack
    while True:
        choice = input("Enter snack code (1-5): ").strip()
        if choice in snacks:
            snack_name, item_price = snacks[choice]
            break
        print("Invalid selection. Please choose a code from 1 to 5.")

    print(f"\nYou selected: {snack_name} (Price: ₹{item_price})")
    print("Please insert coins one at a time.\n")

    # 2. Accept coins one by one until the price is covered
    total_inserted = 0

    while total_inserted < item_price:
        remaining = item_price - total_inserted
        print(f"Inserted: ₹{total_inserted} | Remaining: ₹{remaining}")

        try:
            coin = int(input("Insert coin (₹1, ₹2, ₹5, ₹10, ₹20): "))

            if coin in VALID_COINS:
                total_inserted += coin
                print(f"Accepted ₹{coin} coin.\n")
            else:
                print(f"Invalid coin (₹{coin}). Coin rejected! Please insert a valid coin.\n")

        except ValueError:
            print("Invalid input. Please enter a numeric coin value.\n")

    # 3. Calculate change using the function and dispense snack
    change_owed = calculate_change(total_inserted, item_price)
    dispense_snack(snack_name, item_price, total_inserted, change_owed)


if __name__ == "__main__":
    main()
```
