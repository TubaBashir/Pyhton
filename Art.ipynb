```python
def display_catalog():
    """Displays the available art supplies and returns a price dictionary."""
    catalog = {
        "1": ("Sketchbook (A4)", 250.00),
        "2": ("Acrylic Paint Set (12 Colors)", 480.00),
        "3": ("Fine Tip Brush Pack", 180.00),
        "4": ("Charcoal Pencil Set", 120.00),
        "5": ("Stretched Canvas (12x16 in)", 350.00)
    }

    print("==============================================")
    print("        PALETTE & BRUSH ART SUPPLIES          ")
    print("==============================================")
    print("ID  | Item                             | Price")
    print("----+----------------------------------+------")
    for item_id, (name, price) in catalog.items():
        print(f"{item_id:<3} | {name:<32} | ₹{price:.2f}")
    print("==============================================\n")
    return catalog


def calculate_item_total(price, quantity):
    """Takes unit price and quantity as arguments and returns the line total."""
    return price * quantity


def calculate_tax(subtotal, tax_rate=0.18):
    """Calculates and returns the GST/tax amount based on the subtotal."""
    return subtotal * tax_rate


def calculate_final_total(subtotal, tax_amount):
    """Calculates and returns the final payable amount."""
    return subtotal + tax_amount


def print_bill(customer_name, purchased_items, subtotal, tax_amount, final_total):
    """Prints a complete, formatted bill for the customer."""
    print("\n==============================================")
    print("               OFFICIAL TAX INVOICE           ")
    print("==============================================")
    print(f"Customer Name: {customer_name}")
    print("----------------------------------------------")
    print(f"{'Item':<26} {'Qty':>4} {'Total':>12}")
    print("----------------------------------------------")

    for name, qty, item_total in purchased_items:
        print(f"{name:<26} {qty:>4}   ₹{item_total:>8.2f}")

    print("----------------------------------------------")
    print(f"{'Subtotal:':<32} ₹{subtotal:>8.2f}")
    print(f"{'GST (18%):':<32} ₹{tax_amount:>8.2f}")
    print("==============================================")
    print(f"{'GRAND TOTAL:':<32} ₹{final_total:>8.2f}")
    print("==============================================")
    print(f"Thank you for shopping with us, {customer_name}!")
    print("Keep creating amazing art!")
    print("==============================================\n")


def main():
    # 1. Show catalog and retrieve item data
    catalog = display_catalog()

    customer_name = input("Enter customer name: ").strip() or "Valued Artist"
    purchased_items = []
    subtotal = 0.0

    # 2. Loop to allow purchasing multiple items
    while True:
        choice = input("\nEnter Item ID to add to cart (or 'done' to checkout): ").strip().lower()

        if choice == "done":
            break

        if choice not in catalog:
            print("Invalid Item ID. Please choose a number from 1 to 5.")
            continue

        item_name, unit_price = catalog[choice]

        try:
            quantity = int(input(f"How many '{item_name}' would you like? "))
            if quantity <= 0:
                print("Quantity must be at least 1.")
                continue
        except ValueError:
            print("Invalid input. Please enter a valid whole number.")
            continue

        # 3. Call function with arguments and capture returned value
        line_total = calculate_item_total(unit_price, quantity)
        subtotal += line_total
        purchased_items.append((item_name, quantity, line_total))
        print(f"Added {quantity} x {item_name} (₹{line_total:.2f}) to cart.")

    # 4. Generate bill if items were purchased
    if not purchased_items:
        print("\nNo items purchased. Exiting billing tool.")
        return

    tax_amount = calculate_tax(subtotal)
    final_total = calculate_final_total(subtotal, tax_amount)

    print_bill(customer_name, purchased_items, subtotal, tax_amount, final_total)


if __name__ == "__main__":
    main()
```
