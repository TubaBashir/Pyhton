```python
def calculate_bill(meal_cost, tax_rate, tip_rate):
    """Calculates the total restaurant bill using positional arguments."""
    tax_amount = meal_cost * (tax_rate / 100)
    tip_amount = meal_cost * (tip_rate / 100)
    total_bill = meal_cost + tax_amount + tip_amount
    return round(total_bill, 2)


def seating_arrangements(guests):
    """
    Calculates the total number of unique ways to seat 'guests' in a row
    using recursion (factorial of guests: n!).
    """
    if guests == 0 or guests == 1:
        return 1
    else:
        return guests * seating_arrangements(guests - 1)


def main():
    print("==============================================")
    print("         BILL & SEATING HELPER TOOL           ")
    print("==============================================\n")

    # 1. Access and display the docstring using __doc__
    print("--- Seating Function Documentation ---")
    print(seating_arrangements.__doc__.strip())
    print("----------------------------------------------\n")

    # 2. Get inputs and calculate bill using positional arguments
    try:
        meal_cost = float(input("Enter the base meal cost (₹): "))
        tax_rate = float(input("Enter the tax percentage (%): "))
        tip_rate = float(input("Enter the tip percentage (%): "))

        if meal_cost < 0 or tax_rate < 0 or tip_rate < 0:
            print("Amounts and percentages cannot be negative.")
            return

        # Passing arguments positionally in exact parameter order
        final_bill = calculate_bill(meal_cost, tax_rate, tip_rate)

        # 3. Get guest count and calculate seating arrangements recursively
        guests = int(input("\nEnter the number of guests to seat: "))

        if guests < 0:
            print("Number of guests cannot be negative.")
            return

        arrangements = seating_arrangements(guests)

        # 4. Print complete summary
        print("\n==============================================")
        print("              SUMMARY REPORT                  ")
        print("==============================================")
        print(f"Base Meal Cost        : ₹{meal_cost:.2f}")
        print(f"Tax ({tax_rate}%)            : ₹{meal_cost * (tax_rate / 100):.2f}")
        print(f"Tip ({tip_rate}%)            : ₹{meal_cost * (tip_rate / 100):.2f}")
        print(f"Final Total Bill      : ₹{final_bill:.2f}")
        print("----------------------------------------------")
        print(f"Total Guests          : {guests}")
        print(f"Seating Arrangements  : {arrangements} ways")
        print("==============================================\n")

    except ValueError:
        print("Invalid input. Please enter numeric values for costs and a whole number for guests.")


if __name__ == "__main__":
    main()
```
