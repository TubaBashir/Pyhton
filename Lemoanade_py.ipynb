```python
def greet_customer():
    """Greets the customer and asks for their name."""
    print("========================================")
    print("   WELCOME TO THE FRESH LEMONADE STAND!  ")
    print("========================================")
    name = input("Hello there! What is your name? ").strip()

    # Default to 'Friend' if the user presses Enter without typing a name
    if not name:
        name = "Friend"

    print(f"\nNice to meet you, {name}! Here is our menu:")
    print("- Classic Lemonade: ₹40 per cup\n")
    return name


def calculate_total(cups, price_per_cup):
    """Calculates and returns the total cost of the order."""
    return cups * price_per_cup


def calculate_change(amount_paid, total_cost):
    """Calculates and returns the change due to the customer."""
    return amount_paid - total_cost


def print_receipt(customer_name, cups, price_per_cup, total_cost, amount_paid, change_due):
    """Prints a formatted receipt with a personalized thank-you message."""
    print("\n========================================")
    print("            OFFICIAL RECEIPT            ")
    print("========================================")
    print(f"Customer:      {customer_name}")
    print(f"Item:          Fresh Lemonade")
    print(f"Quantity:      {cups} cup(s) @ ₹{price_per_cup:.2f}")
    print("----------------------------------------")
    print(f"Total Cost:    ₹{total_cost:.2f}")
    print(f"Amount Paid:   ₹{amount_paid:.2f}")
    print(f"Change Due:    ₹{change_due:.2f}")
    print("========================================")
    print(f"Thank you for your business, {customer_name}!")
    print("Stay cool and have a wonderful day!")
    print("========================================\n")


def main():
    PRICE_PER_CUP = 40.00  # Price per cup in INR

    # 1. Greet customer and capture their name
    customer_name = greet_customer()

    # 2. Get the order quantity
    while True:
        try:
            cups = int(input(f"How many cups of lemonade would you like, {customer_name}? "))
            if cups > 0:
                break
            print("Please enter at least 1 cup.")
        except ValueError:
            print("Invalid input. Please enter a whole number.")

    # 3. Calculate total using arguments and return value
    total_cost = calculate_total(cups, PRICE_PER_CUP)
    print(f"\nYour total comes to: ₹{total_cost:.2f}")

    # 4. Get payment and ensure it covers the total cost
    while True:
        try:
            amount_paid = float(input("Enter the cash amount paid: ₹"))
            if amount_paid >= total_cost:
                break
            print(f"Insufficient amount. You need at least ₹{total_cost:.2f}.")
        except ValueError:
            print("Invalid input. Please enter a valid numeric amount.")

    # 5. Calculate change using arguments and return value
    change_due = calculate_change(amount_paid, total_cost)

    # 6. Print the final receipt and personalized message
    print_receipt(customer_name, cups, PRICE_PER_CUP, total_cost, amount_paid, change_due)


# Run the program
if __name__ == "__main__":
    main()
```
