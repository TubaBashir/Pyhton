```python
def calculate_change(total_paid, ticket_cost):
    """Calculates and returns the change owed after paying the parking ticket."""
    return total_paid - ticket_cost


def main():
    TICKET_COST = 30
    VALID_COINS = [1, 2, 5, 10, 20]
    total_paid = 0

    print("========================================")
    print("     PARKING TICKET PAYMENT HELPER      ")
    print("========================================")
    print(f"Parking Ticket Fee : ₹{TICKET_COST}")
    print("Accepted Coins     : ₹1, ₹2, ₹5, ₹10, ₹20\n")

    # Collect coins until the ticket fee is fully covered
    while True:
        remaining = TICKET_COST - total_paid
        print(f"Paid: ₹{total_paid} | Remaining: ₹{remaining}")

        try:
            coin = int(input("Insert coin: ₹"))
        except ValueError:
            print("Invalid input. Please enter a whole number.\n")
            continue

        # Skip invalid coins using 'continue'
        if coin not in VALID_COINS:
            print(f"Rejected ₹{coin}. Only ₹1, ₹2, ₹5, ₹10, or ₹20 coins are accepted.\n")
            continue

        total_paid += coin
        print(f"Accepted ₹{coin} coin.\n")

        # Stop coin collection using 'break' once enough money is inserted
        if total_paid >= TICKET_COST:
            break

    # Calculate change using the function
    change = calculate_change(total_paid, TICKET_COST)

    print("========================================")
    print("            PAYMENT RECEIPT             ")
    print("========================================")
    print(f"Ticket Cost   : ₹{TICKET_COST}")
    print(f"Total Paid    : ₹{total_paid}")

    # Use 'pass' when exact payment is made and no change is needed
    if change == 0:
        pass
    else:
        print(f"Change Due    : ₹{change}")

    print("----------------------------------------")
    print("Ticket validated. Safe travels!")
    print("========================================\n")


if __name__ == "__main__":
    main()
```
