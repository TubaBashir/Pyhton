```python
def cube(number):
    """Calculates and returns the cube of a given number."""
    return number ** 3


def by_three(number):
    """
    Checks if a number is divisible by 3.
    If it is, executes and returns the result of cube(number).
    Otherwise, returns False.
    """
    if number % 3 == 0:
        return cube(number)
    else:
        return False


def main():
    print("--- Cube of the Cube Checker ---")

    # Get user input with error handling
    try:
        num = int(input("Enter a whole number: "))
        result = by_three(num)

        if result is not False:
            print(f"{num} is divisible by 3! Its cube is: {result}")
        else:
            print(f"{num} is NOT divisible by 3, so the cube function was not executed.")

    except ValueError:
        print("Invalid input. Please enter a valid integer.")


# Run the program
if __name__ == "__main__":
    main()
```
